# RadImageNet-initialized ResNet-50 on the fixed HPO-v1 fold

This is a single controlled comparison against the HPO-v1 +281 train-only ResNet-50 recipe. The only intended change is the image trunk initialization: the official RadImageNet ResNet-50 weights replace ImageNet weights. The 12-target head and optimizer start fresh.

The run uses the same 24 slice windows per study, effective study batch 8, training labels, +281 training-only studies, and fixed 825-study report validation scored against Steven v4. The 58 gold studies are excluded from training and validation.

Inputs are fixed paths in the code; there is no recursive file search. Each epoch writes best/latest checkpoints to `/kaggle/working/radimagenet_resnet50_fold0`; a snapshot copy is refreshed every two epochs. Kaggle exposes these files as notebook outputs after the run completes. No API token is needed.

```python
from pathlib import Path
import runpy
import os

# Kaggle materializes this archive as the nested dataset path shown in `kaggle datasets files`.
asset_root = Path('/kaggle/input/datasets/tumis001/rsna-hpo-blend-radimagenet-kaggle-assets-v1/radimagenet_assets')
runner = asset_root / 'source/run_kaggle_radimagenet.py'
if not runner.is_file():
    raise FileNotFoundError(f'Fixed experiment runner missing: {runner}')
if not (asset_root / 'weights/radimagenet_resnet50.pt').is_file():
    raise FileNotFoundError(f'Fixed RadImageNet weights missing under {asset_root}')
os.environ['RSNA_RADIMAGENET_ASSET_ROOT'] = str(asset_root)
runpy.run_path(str(runner), run_name='__main__')
```
